# Assignment:
### Create a plot comparing Kohli, Rohit Sharma, and Sehwag across 10 years of hypothetical runs.
#### Use:
- Labels
- Legends
- Colors
- Line styles
- One custom style

```python
# Sample Data
# An array for a 10-year period to serve as our x-axis.
years = np.arange(2015, 2025)

# Hypothetical runs for each player. I've tried to make the numbers
# reflect their general playing styles for fun.
# Kohli: The model of consistency.
kohli_runs = np.array([1100, 1450, 1050, 1320, 1280, 970, 1150, 1080, 1370, 1190])
# Rohit Sharma: Capable of huge peaks.
rohit_runs = np.array([850, 990, 1290, 740, 1490, 640, 1120, 890, 1250, 1030])
# Sehwag: Fearless and aggressive (hypothetical data as his peak was earlier).
sehwag_runs = np.array([1050, 650, 1180, 1420, 780, 1230, 810, 950, 620, 1090])


plt.plot(years, kohli_runs, label="kohli", linestyle="--")
plt.plot(years, rohit_runs, label="rohit", linestyle=":")
plt.plot(years, sehwag_runs, label="sehwag", linestyle="-")

plt.xlabel("years")
plt.ylabel("runs")

plt.legend()
plt.style.use("seaborn-v0_8-bright")
plt.grid()
plt.tight_layout()

plt.show()
```
