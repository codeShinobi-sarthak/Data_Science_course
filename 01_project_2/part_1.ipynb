```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c) > 3]
print(chunks[0])
```

```python
chunks
```

```python
def parse_number(s):
    """Convert short number formats like '12.5K', '1.2M' to int."""
    if s is None:
        return None
    s = s.strip().replace(',', '').upper()
    try:
        if s.endswith('K'):
            return int(float(s[:-1]) * 1_000)
        if s.endswith('M'):
            return int(float(s[:-1]) * 1_000_000)
        if s.endswith('B'):
            return int(float(s[:-1]) * 1_000_000_000)
        return int(float(s))
    except Exception:
        return None
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    # Safety check: if chunk is empty, skip it
    if not chunk: 
        return None 
        
    lines = chunk.split("\n")
    
    handle = lines[0] 
    posts = parse_number(lines[1].split()[0])
    followers = parse_number(lines[2].replace(" followers", ""))
    following = parse_number(lines[3].replace(" following", ""))
    

    display_name = ""
    if len(lines) > 4:
        display_name = lines[4]

    # Line 5: Type of Page (e.g., "Nonprofit organization")
    type_of_page = 'unknown'
    if len(lines) > 5:
        type_of_page = lines[5]
        
    # Line 6: Bio (e.g., "The association...")
    bio = ""
    if len(lines) > 6:
        bio = lines[6]

    return {
        "name": handle,          # The unique username
        "display_name": display_name, # The name shown on profile
        "posts": posts,
        "followers": followers,
        "following": following,
        "type": type_of_page,
        "bio": bio
    }
```

```python
parse_chunk(chunks[0])
```

```python
all_chunks = []
for chunk in chunks:
    c = parse_chunk(chunk)
    all_chunks.append(c)
print(all_chunks)
```

```python
import json
```

```python
s = json.dumps(all_chunks, indent=4)
with open("json.data", "w") as f:
    f.write(s)
print(s)
```

##  Starting with answering the questions 

- Who has the maximum posts
- who has the maximum followers
- Who follows the maximum people
- How many categories (Digital creators, Non-profit foundation, etc do we have? How many people do we have?


```python
max_posts = -1  
who_posts = {}

for c in all_chunks:
    if c['posts'] > max_posts:
        max_posts = c['posts']
        who_posts = c

print(f"Most posts: {who_posts}")
```

```python
max_followers = -1
who_followers = {}

for c in all_chunks:
    if c['followers'] > max_followers:
        max_followers = c['followers']
        who_followers = c

print(f"Most followers: {who_followers}")
```

```python
max_following = -1
who_following = {}

for c in all_chunks:
    if c['following'] > max_following:
        max_following = c['following']
        who_following = c

print(f"Follows most people: {who_following}")
```

```python
categories = set()
for chunk in all_chunks:
    categories.add(chunk['type'])
    
print(categories)
print(f"categories count : {len(categories)}")
```
