```python
import json


def load_data(file_name):
    with open(file_name, "r") as f:
        return json.load(f)


def find_people(user_id, data):
    user_friends = {}
    
    # Mapping user IDs to their friends (as a set for faster lookup)
    for user in data['users']:
        user_friends[user['id']] = set(user['friends'])
        
    # If the user ID is not in the data, return an empty list
    if user_id not in user_friends:
        return []
        
    direct_friends = user_friends[user_id]
    
    suggestions = {}
    # For each of the user's direct friends
    for friend in direct_friends:
        for mutual in user_friends[friend]:
            if mutual != user_id and mutual not in direct_friends:
                # Count mutual friends
                suggestions[mutual] = suggestions.get(mutual, 0) + 1

    # Sort the suggestions by the count of mutual friends in descending order
    sorted_suggestions = sorted(suggestions.items(), key=lambda x: x[1], reverse=True)
    
    # Return only the user IDs (not mutual friend counts)
    return [user_id for user_id, mutual_count in sorted_suggestions]


# Load data from the file
data = load_data("data.json")

# Set the user ID you want to find recommendations for
user_id = 10

# Get the friend recommendations
recommendation = find_people(user_id, data)

# Print the recommendations
print(recommendation)
```
