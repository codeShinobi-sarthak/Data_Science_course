```python
import json
```

```python
# function to load data
def load_data(file_name):
    with open(file_name, "r") as file:
        data = json.load(file)
    return data
```

```python
data = load_data("data.json")
```

```python
def display_user(data):
    for user in data['users']:
        print(f"{user['name']} is friends with: {user['friends']}")

display_user(data);
```
