```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

```python
# Load the data
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize the input data
x_train = x_train / 255.0
x_test = x_test / 255.0

# One-hot encode the labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
# Build the Neural Network structure 🏗️
model = Sequential([
  Flatten(input_shape=(28, 28)), # Flattens the 28x28 image into a single line
  Dense(128, activation='relu'), # A hidden layer with 128 neurons
  Dense(10, activation='softmax') # Output layer (digits 0-9)
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
print("Training the model")
model.fit(x_train, y_train, epochs=5, batch_size=32)
```

```python
print("Testing the model")
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"test_loss : {test_loss:.4f}")
print(f"test_acc : {test_acc:.4f}")
```
