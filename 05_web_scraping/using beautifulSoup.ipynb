```python
from bs4 import BeautifulSoup
import requests
```

```python
with open("html/pages.html") as f:
    content = f.read()
```

```python
soup = BeautifulSoup(content, "html.parser")
```

```python
soup.h1
```

```python
all_h3 = soup.find_all("h3")
```

```python
all_h3
```

```python
articles = soup.select("article.product_pod")
```

```python
items = []
items2 = []

for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text.split("£")[1]
    # items.append(f"{title} : {price}")
    items2.append([title, price])
```

```python
items2
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(items2, columns=["Book Title", "prices"])
print(df)
```
