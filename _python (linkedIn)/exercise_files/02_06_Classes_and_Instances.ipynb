## Classes

```python
class Dog:
    def __init__(self, name):
        self.name = name
        self.legs = 4
    
    def speak(self):
        print(self.name + ' says: Bark!')
        
```

```python
my_dog = Dog('Rover')
another_dog = Dog('Fluffy')
```

```python
my_dog.speak()
```

```python
another_dog.speak()
```

```python
class Hero:

    # Instance method — needs 'self'
    def attack(self):
        print("Hero attacks!")

    # Static method — no 'self'
    @staticmethod
    def shout():
        print("Aaaaah!")

    # Class method — uses 'cls'
    @classmethod
    def info(cls):
        print("I belong to the Hero class.")
```

```python
h = Hero()
```

```python
h.attack()
Hero.attack(h)
```

```python
h.shout()
Hero.shout()
```

```python
h.info()
Hero.info()
```
