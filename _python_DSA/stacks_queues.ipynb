## stacks using deque from collection

```python
from collections import deque

# 1. Initialization
stack = deque()

# 2. Push (add to top) - Use append() (O(1))
stack.append('A')
stack.append('B')
stack.append('C')
print(f"Stack: {stack}")

# 3. Pop (remove from top) - Use pop() (O(1))
top_element = stack.pop()
print(f"Popped: {top_element}, Stack is now: {stack}") # Popped C

top_element = stack.pop()
print(f"Popped: {top_element}, Stack is now: {stack}") # Popped B
```

## Queues - First in First out

```python
from collections import deque

q = deque()

print(q)
```

```python
#  Enqueue  - will add elements to the right
q.append(5)
q.append(6)
q.append(10)
q.append(22)
q.append(7)
q.append(9)

q
```

```python
#  Deque (pop left) - this is remove the elm from the left O(1)
#! writing left is important 
#  but we will see both 


print(f"usinf only pop() : {q.pop()}")
print(f"usinf only popleft() : {q.popleft()}")
```

```python
# empty the queue
while q:
    print(q.popleft())
```

## lets now have a look on sets

```python
# 1. Initialization
# Create an empty set - MUST use set()
my_set = set()
# Create a set with initial elements
# Duplicates are automatically removed
numbers_set = {1, 2, 3, 3, 4, 4, 4}
print(f"Initial set: {numbers_set}")

# 2. Adding Elements (Average O(1))
numbers_set.add(5)
numbers_set.add(1) # Does nothing, 1 is already in the set
print(f"After adding elements: {numbers_set}")

# 3. Removing Elements (Average O(1))
# .remove() will raise a KeyError if the element doesn't exist
numbers_set.remove(3)
print(f"After removing 3: {numbers_set}")

# .discard() will NOT raise an error if the element is not found
numbers_set.discard(99) # Does nothing, no error
print(f"After discarding 99: {numbers_set}")

# 4. Membership Testing (Average O(1))
# This is the primary strength of a set!
print(f"Is 4 in the set? {4 in numbers_set}")
print(f"Is 99 in the set? {99 in numbers_set}")
```
