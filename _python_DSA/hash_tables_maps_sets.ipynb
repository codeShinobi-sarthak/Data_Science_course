## Hash funtions

Now what is `hashable` ?
- strings
- integers
- tuples (1, 2, 3)

basically `immutable` objects are hashable thus `array`, `dict` are `not hashable`

```python
# The size of our hash map array
SIZE = 5 
# We'll use a list of lists to build the map, which makes handling collisions easy
hash_map = [[] for _ in range(SIZE)]

# Its only job is to return an index.
def hash_function(data):
    """
    Calculates a hash index for a given string.
    """
    total = 0
    # Use ord() to get the numerical value of characters for calculations
    for char in data:
        total += ord(char)
        
    # The modulo operator gives us an index within the bounds of our map size
    return total % SIZE

# --- An Insert Function ---
# This function uses the hash function to place data in the map.
def insert(data):
    """
    Hashes the data and inserts it into the correct spot in the hash map.
    """
    # 1. Get the index from the hash function
    index = hash_function(data)
    
    # 2. Add the data to the list at that index (handles collisions)
    hash_map[index].append(data)

```

```python
# --- Let's see it in action! ---
insert("apple")
insert("banana")
insert("grape")
insert("orange")
insert("melon") # This will likely cause a collision
insert("peach") 

print("The final hash map:")
print(hash_map)

print(f"\n'apple' is at index: {hash_function('apple')}")
print(f"'peach' is at index: {hash_function('peach')}")

# here you see the same idx
print(f"'melon' is at index: {hash_function('melon')}")
print(f"'banana' is at index: {hash_function('banana')}")
```

```python
#  this is to empty the map 
def empty_map():
    hash_map.clear()

empty_map()
```

## Below lets have code for python hashmap using dicionary

```python
# 1. Create an empty dictionary (hash map)
# This is our container for key-value pairs.
phone_book = {}
print(f"Empty Phone Book: {phone_book}")

# 2. Add items (mapping keys to values)
# Python automatically hashes the key ('name') to store the value ('number').
print("\nAdding entries...")
phone_book['Jenny'] = 8675309
phone_book['Emergency'] = 911
phone_book['Alice'] = 1234567
print(f"Updated Phone Book: {phone_book}")

# 3. Access a value using its key
# This is a very fast operation, the main advantage of a hash map.
print(f"\nAccessing Jenny's number: {phone_book['Jenny']}")

# 4. Update an existing value
# Just assign a new value to the existing key.
print("\nUpdating Alice's number...")
phone_book['Alice'] = 7654321
print(f"After update: {phone_book}")

# 5. Check if a key exists
# Use the 'in' keyword to avoid errors when accessing a non-existent key.
print(f"\nIs 'Bob' in the phone book? {'Bob' in phone_book}")
print(f"Is 'Jenny' in the phone book? {'Jenny' in phone_book}")

# 6. Remove an item
# Use the 'del' keyword to remove a key-value pair.
print("\nDeleting 'Jenny'...")
del phone_book['Jenny']
print(f"Final Phone Book: {phone_book}")
```

## hash sets

```python
# Empty set
my_hash_set = set()

# Set with initial elements
another_hash_set = {1, 2, 3, "apple"}
```

```python
my_hash_set.add(10)
my_hash_set.add("banana")
my_hash_set.add(10) # This will not add a duplicate

print(my_hash_set)
```

```python
my_hash_set.remove(10)

# Using discard() to avoid KeyError
my_hash_set.discard("orange") # Will not raise an error if "orange" is not present

print(my_hash_set)
```

# Default dict

```python
from collections import defaultdict

# The "new way" with defaultdict
word_list = ["apple", "banana", "apple", "orange", "banana", "apple"]

# Provide int as the default factory. int() returns 0.
word_counts = defaultdict(int) 

for word in word_list:
    # No check needed! 
    # If 'word' is new, word_counts[word] is set to 0, then incremented.
    word_counts[word] += 1


#  okk this is something to look at
print(f"does it exists : {word_counts[3]}") 

print(f"defaultdict result: {dict(word_counts)}") # Convert to dict for clean printing
```
