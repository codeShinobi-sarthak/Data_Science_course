## eaps / Priority Queues (heapq)

Python provides the heapq module, which implements a `min-heap`. It's `not` a `class`; it's a `set of functions` that you apply to a standard Python list.

```python
import heapq
```

```python
# 1. Initialization
# A heap is just a list that you maintain with heapq functions
min_heap = []

# 2. Push elements onto the heap (O(log n))
heapq.heappush(min_heap, 4)
heapq.heappush(min_heap, 1)
heapq.heappush(min_heap, 7)
heapq.heappush(min_heap, 3)


# The list might not look sorted, but it satisfies the heap property
print(f"heap list : {min_heap}")
```

```python
# 3. Pop the smallest element (O(log n))
smallest = heapq.heappop(min_heap)
print(f"Popped smallest element: {smallest}, Heap is now: {min_heap}")

smallest = heapq.heappop(min_heap)
print(f"Popped smallest element: {smallest}, Heap is now: {min_heap}")
```

### Now lets have a look on to how to make a max heap

Tip: To simulate a max-heap, push and pop the negative of the values.

```python
max_heap = []

heapq.heappush(max_heap, -4)
heapq.heappush(max_heap, -1)
heapq.heappush(max_heap, -7)

largest = -heapq.heappop(max_heap)  # Pop -7, negate to get 7
print(f"Largest element from max-heap: {largest}")
```
