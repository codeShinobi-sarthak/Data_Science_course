```python
"""
This graph representation is good if
you want to represent it with weights like this

A -----> B (weight = 3)
A -----> C (weight = 7)
c -----> D (weight = 10)

OR 
you can even do it like this and store as a tuple or list of list

 (A, B, 3)
 (A, C, 7)
 (C, D, 10)
 
"""

# Mapping idx: A=0, B=1, C=2, D=3
# Using infinity to represent no direct path
inf = float('inf')

# Rows represent the 'from' node, columns are the 'to' node
adjacency_matrix = [
    #   A    B    C     D
    [ inf,   3,   7,   inf ],  # From A
    [ inf, inf, inf,   inf ],  # From B
    [ inf, inf, inf,    10 ],  # From C
    [ inf, inf, inf,   inf ]   # From D
]

print(f"adjacency_matrix below : \n  {adjacency_matrix} \n")

# To find the weight of the edge from A (0) to C (2):
weight_A_C = adjacency_matrix[0][2]
print(f"Weight from A to C is: {weight_A_C} \n")
# Output: Weight from A to C is: 7.0

edge_list = [
    ('A', 'B', 3),
    ('A', 'C', 7),
    ('C', 'D', 10)
]

# To work with the edges, you just iterate through the list:
for source, destination, weight in edge_list:
    print(f"Edge from {source} to {destination} with weight {weight}")
```

## now we will move to dfs and bfs

```python
# We'll represent it as a dictionary where keys are nodes
# and values are lists of their neighbors and is a unweighted graph fro simplicity
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': [],
    'F': ['H'],
    'G': ['H'],
    'H': []
}
```

```python
def dfs(visited, graph, node):
    """
    Performs a Depth-First Search traversal on a graph.
    """
    if node not in visited:
        # Process the node (in this case, just print it)
        print(node, end=' ')
        
        # Mark the node as visited
        visited.add(node)
        
        # Recursively visit all the neighbors
        for neighbor in graph[node]:
            dfs(visited, graph, neighbor)

# --- How to run it ---
print("DFS Traversal:")
visited_dfs = set()  # Create a set to store visited nodes
dfs(visited_dfs, graph, 'A') # Start the traversal from node 'A'
print("\n") 
# Expected Output: A B E C F H G D (your exact order might vary for neighbors)
```

```python
import collections

def bfs(graph, start_node):
    """
    Performs a Breadth-First Search traversal on a graph.
    """
    visited = {start_node} # A set to keep track of visited nodes
    queue = collections.deque([start_node]) # A queue for nodes to visit

    while queue:
        # Dequeue a vertex from the beginning of the list
        node = queue.popleft()
        print(node, end=' ')

        # Get all adjacent vertices of the dequeued vertex.
        # If an adjacent has not been visited, then mark it
        # visited and enqueue it.
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

# --- How to run it ---
print("BFS Traversal:")
bfs(graph, 'A') # Start the traversal from node 'A'
print()
# Expected Output: A B C D E F G H
```
