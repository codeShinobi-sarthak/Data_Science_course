## Single Linked List

```python
class Node:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next

    def __str__(self):
        return str(self.val)
```

```python
head = Node(4)
node_1 = Node(5)
node_2 = Node(8)

head.next = node_1
node_1.next = node_2
```

```python
print(head)
```

```python
temp = head

while temp:
    print(f"{temp}  ---> ", end = "")
    temp = temp.next
print("null")
```

## Double Linked List

```python
class Node:
    def __init__(self, val, next=None, prev=None):
        self.val = val
        self.next = next
        self.prev = prev

    def __str__(self):
        return str(self.val)
```

```python
head = Node(4)
node_1 = Node(5)
node_2 = Node(8)

head.next = node_1
node_1.next = node_2

node_1.prev = head
node_2.prev = node_1
```

```python
temp = head

print("------printing using next pointer-------")
while temp:
    print(f"{temp}  ---> ", end = "")
    temp = temp.next
print("null")



print("\n------printing using prev pointer-------")

#  find the very last node (the tail)
last_node = head
if last_node: 
    while last_node.next:
        last_node = last_node.next

# 2. Now, start from the last node and move backward using .prev
temp = last_node
while temp:
    print(f"{temp.val} ---> ", end="") 
    temp = temp.prev 
print("null")
```
