```python
# 1. The class definition for a single tree node
class TreeNode:
    """A node in a binary tree."""
    def __init__(self, value=0, left_child=None, right_child=None):
        self.val = value
        self.left = left_child
        self.right = right_child

    # 2. A function to insert a new value into the BST
    def insert(root, value):
        """
        Inserts a new value into the binary search tree.
        Returns the root of the modified tree.
        """
        # If the tree/subtree is empty, create a new node and return it
        if not root:
            return TreeNode(value)
        
        # If the value is less than the current node's value, go to the left subtree
        if value < root.val:
            root.left = insert(root.left, value)
        # Otherwise, go to the right subtree
        else:
            root.right = insert(root.right, value)
            
        return root
    
    # 3. A function to print the tree in-order (left, root, right)
    def print_in_order(node):
        """
        Prints the values of the tree by traversing in-order.
        This will print the values of a BST in sorted order.
        """
        if not node:
            return
        
        print_in_order(node.left)
        print(node.val, end=" ") # Print the value of the current node
        print_in_order(node.right)

    def __repr__(self):
        return f"{self.val}"
    


```

```python
# --- Main script to build and print the tree ---

# The values you want to add to the tree
values = [10, 7, 1, 0, 3, 6, 2]

# The root of our tree starts as empty
root = None

# Loop through each value and insert it into the tree
for value in values:
    root = insert(root, value)
    
print("Tree constructed successfully!")

# Let's print the tree's values to verify the structure
print("In-order traversal of the tree:")
print_in_order(root)
# Expected output: 0 1 2 3 6 7 10
```

```python
from collections import deque

def level_order(node):
    print("Level-order traversal:", end="\n\n")
    q = deque()
    q.append(node)
    lev = 0;
    
    while q:
        s = len(q)
        print(f"level {lev} --> ", end="")
        for _ in range(s):
            node = q.popleft()
            print(node, end=" ")
    
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        print()
        lev += 1
        

level_order(root)
    
```
